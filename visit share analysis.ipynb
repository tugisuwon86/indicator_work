```python
# =========================================================
# CONFIG
# =========================================================

STORE_ID = "id"
STORE_TYPE = "name"
MONTH = "month"
LAT = "lat"
LON = "lng"
VISITS = "visits"
DMA = "dma"
SUBMARKET = "submarket"

OUR_STORE_TYPE = "Chick-fil-A"

RADIUS_LIST = [1, 3, 5, 10]
EVENT_PRE_MONTHS = 6
EVENT_POST_MONTHS = 12
```

```python
def load_store_data(path):
    df = pd.read_parquet(path)
    df[MONTH] = df['start_date'].apply(lambda x: x.split('/')[-1] + '-' + x.split('/')[0].zfill(2) + '-' + x.split('/')[1].zfill(2))
    df[MONTH] = pd.to_datetime(df[MONTH])
    df[MONTH] = df[MONTH].dt.to_period("M").dt.to_timestamp()
    
    df = df.dropna(subset=[STORE_ID, STORE_TYPE, MONTH, LAT, LON, VISITS])
    
    df[VISITS] = pd.to_numeric(df[VISITS], errors="coerce")
    df = df.dropna(subset=[VISITS])
    return df
```

```python
# =========================================================
# 2. HAVERSINE DISTANCE
# =========================================================

def haversine_miles(lat1, lon1, lat2, lon2):
    r = 3958.8  # Earth radius in miles

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return r * c
```

```python
# =========================================================
# 3. DETECT NEW OPENINGS
# =========================================================

def detect_openings(df):
    first_data_month = df[MONTH].min()

    store_master = (
        df.sort_values(MONTH)
        .groupby(STORE_ID)
        .agg(
            store_type=(STORE_TYPE, "first"),
            opening_month=(MONTH, "min"),
            lat=(LAT, "first"),
            lon=(LON, "first"),
            dma=(DMA, "first"),
            submarket=(SUBMARKET, "first"),
        )
        .reset_index()
    )

    store_master["is_new_store"] = store_master["opening_month"] > first_data_month
    store_master["is_our_store"] = store_master["store_type"] == OUR_STORE_TYPE
    store_master["is_competitor"] = store_master["store_type"] != OUR_STORE_TYPE

    return store_master
```

```python
# =========================================================
# 4. BUILD OUR STORE MONTHLY PANEL
# =========================================================

def build_our_store_panel(df, store_master):
    our_ids = store_master.loc[store_master["is_our_store"], STORE_ID]

    panel = df[df[STORE_ID].isin(our_ids)].copy()

    panel = panel.rename(
        columns={
            STORE_ID: "our_store_id",
            STORE_TYPE: "our_store_type",
            LAT: "our_lat",
            LON: "our_lon",
            VISITS: "our_visits",
            DMA: "dma",
            SUBMARKET: "submarket",
        }
    )

    panel = panel[
        [
            "our_store_id",
            MONTH,
            "our_store_type",
            "our_lat",
            "our_lon",
            "our_visits",
            "dma",
            "submarket",
        ]
    ]

    return panel
```

```python
# =========================================================
# 5. SPATIAL FEATURES FOR NEW OPENINGS
# =========================================================

def add_opening_features(panel, store_master, geo_unit="submarket"):
    """
    geo_unit can be:
    - "submarket"
    - "dma"
    """

    results = []

    openings = store_master[store_master["is_new_store"]].copy()

    for _, row in panel.iterrows():
        month = row[MONTH]
        our_store_id = row["our_store_id"]
        our_lat = row["our_lat"]
        our_lon = row["our_lon"]
        geo_value = row[geo_unit]

        candidates = openings[
            (openings[geo_unit] == geo_value)
            & (openings[STORE_ID] != our_store_id)
            & (openings["opening_month"] <= month)
            & (
                openings["closing_month"].isna()
                | (openings["closing_month"] > month)
            )
        ].copy()

        base = row.to_dict()

        for r in RADIUS_LIST:
            base[f"new_comp_{r}mi"] = 0
            base[f"new_our_{r}mi"] = 0

        base["nearest_new_comp_dist"] = np.nan
        base["nearest_new_our_dist"] = np.nan
        base["new_comp_weighted_score"] = 0.0
        base["new_our_weighted_score"] = 0.0

        if len(candidates) > 0:
            candidates["distance_miles"] = haversine_miles(
                our_lat,
                our_lon,
                candidates["lat"].values,
                candidates["lon"].values,
            )

            comp = candidates[candidates["is_competitor"]]
            our = candidates[candidates["is_our_store"]]

            for r in RADIUS_LIST:
                base[f"new_comp_{r}mi"] = (comp["distance_miles"] <= r).sum()
                base[f"new_our_{r}mi"] = (our["distance_miles"] <= r).sum()

            if len(comp) > 0:
                base["nearest_new_comp_dist"] = comp["distance_miles"].min()
                base["new_comp_weighted_score"] = (
                    1 / comp["distance_miles"].clip(lower=0.1)
                ).sum()

            if len(our) > 0:
                base["nearest_new_our_dist"] = our["distance_miles"].min()
                base["new_our_weighted_score"] = (
                    1 / our["distance_miles"].clip(lower=0.1)
                ).sum()

        results.append(base)

    return pd.DataFrame(results)
```

```python
def add_consecutive_flags(g, value_col="visits"):
    g = g.sort_values(MONTH).copy()

    g["is_nonzero"] = g[value_col].fillna(0) > 0
    g["is_zero"] = g[value_col].fillna(0) <= 0

    g["nonzero_streak"] = (
        g["is_nonzero"]
        .groupby((g["is_nonzero"] != g["is_nonzero"].shift()).cumsum())
        .cumcount() + 1
    )

    g.loc[~g["is_nonzero"], "nonzero_streak"] = 0

    g["zero_streak"] = (
        g["is_zero"]
        .groupby((g["is_zero"] != g["is_zero"].shift()).cumsum())
        .cumcount() + 1
    )

    g.loc[~g["is_zero"], "zero_streak"] = 0

    return g


def detect_openings_and_closings(df):
    first_data_month = df[MONTH].min()

    df2 = (
        df.groupby(STORE_ID, group_keys=False)
        .apply(add_consecutive_flags, value_col=VISITS)
        .reset_index(drop=True)
    )

    store_rows = []

    for store_id, g in df2.groupby(STORE_ID):
        g = g.sort_values(MONTH).copy()

        store_type = g[STORE_TYPE].iloc[0]
        lat = g[LAT].iloc[0]
        lon = g[LON].iloc[0]
        dma = g[DMA].iloc[0]
        submarket = g[SUBMARKET].iloc[0]

        # opening = first month of a 3-month nonzero streak
        opening_candidates = g[g["nonzero_streak"] >= 3]

        if len(opening_candidates) > 0:
            third_month = opening_candidates.iloc[0][MONTH]
            opening_month = third_month - pd.DateOffset(months=2)
        else:
            opening_month = pd.NaT

        # closing = first month of a 6-month zero streak
        closing_candidates = g[g["zero_streak"] >= 6]

        if len(closing_candidates) > 0:
            sixth_zero_month = closing_candidates.iloc[0][MONTH]
            closing_month = sixth_zero_month - pd.DateOffset(months=5)
        else:
            closing_month = pd.NaT

        is_new_store = (
            pd.notna(opening_month)
            and opening_month > first_data_month
        )

        is_closed = pd.notna(closing_month)

        store_rows.append({
            STORE_ID: store_id,
            "store_type": store_type,
            "opening_month": opening_month,
            "closing_month": closing_month,
            "is_new_store": is_new_store,
            "is_closed": is_closed,
            "lat": lat,
            "lon": lon,
            "dma": dma,
            "submarket": submarket,
            "is_our_store": store_type == OUR_STORE_TYPE,
            "is_competitor": store_type != OUR_STORE_TYPE,
        })

    store_master = pd.DataFrame(store_rows)

    return store_master, df2
```

```python
# =========================================================
# 6. ACTIVE POST-OPENING FEATURES
# =========================================================

def add_active_opening_features(panel, store_master, geo_unit="submarket"):
    """
    Counts stores that opened historically and remain active after opening.
    Useful for estimating cumulative impact.
    """

    results = []

    openings = store_master[store_master["is_new_store"]].copy()

    for _, row in panel.iterrows():
        month = row[MONTH]
        our_store_id = row["our_store_id"]
        our_lat = row["our_lat"]
        our_lon = row["our_lon"]
        geo_value = row[geo_unit]

        candidates = openings[
            (openings[geo_unit] == geo_value)
            & (openings[STORE_ID] != our_store_id)
            & (openings["opening_month"] <= month)
        ].copy()

        base = row.to_dict()

        for r in RADIUS_LIST:
            base[f"active_comp_{r}mi"] = 0
            base[f"active_our_{r}mi"] = 0

        base["active_comp_weighted_score"] = 0.0
        base["active_our_weighted_score"] = 0.0

        if len(candidates) > 0:
            candidates["distance_miles"] = haversine_miles(
                our_lat,
                our_lon,
                candidates["lat"].values,
                candidates["lon"].values,
            )

            comp = candidates[candidates["is_competitor"]]
            our = candidates[candidates["is_our_store"]]

            for r in RADIUS_LIST:
                base[f"active_comp_{r}mi"] = (comp["distance_miles"] <= r).sum()
                base[f"active_our_{r}mi"] = (our["distance_miles"] <= r).sum()

            if len(comp) > 0:
                base["active_comp_weighted_score"] = (
                    1 / comp["distance_miles"].clip(lower=0.1)
                ).sum()

            if len(our) > 0:
                base["active_our_weighted_score"] = (
                    1 / our["distance_miles"].clip(lower=0.1)
                ).sum()

        results.append(base)

    return pd.DataFrame(results)
```

```python
def build_event_catalog(
    analysis_df,
    store_master,
    geo_unit="submarket",
    radius=10,
    pre_months=3,
    post_months=3,
):

    events = []

    our_stores = store_master[
        store_master.is_our_store
    ]

    new_events = store_master[
        store_master.is_new_store
    ]

    event_id = 0

    for _, event in new_events.iterrows():

        event_id += 1

        event_store = event[STORE_ID]
        event_month = event["opening_month"]

        event_lat = event["lat"]
        event_lon = event["lon"]

        event_type = (
            "Competitor Open"
            if event["is_competitor"]
            else "Our Open"
        )

        candidates = our_stores[
            our_stores[geo_unit] == event[geo_unit]
        ].copy()

        candidates = candidates[
            candidates[STORE_ID] != event_store
        ]

        if len(candidates) == 0:
            continue

        candidates["distance"] = haversine_miles(
            candidates["lat"].values,
            candidates["lon"].values,
            event_lat,
            event_lon,
        )

        candidates = candidates[
            candidates.distance <= radius
        ]

        for _, affected in candidates.iterrows():

            affected_store = affected[STORE_ID]

            hist = analysis_df[
                analysis_df[STORE_ID] == affected_store
            ].copy()

            hist = hist.sort_values(MONTH)

            pre = hist[
                (hist[MONTH] >= event_month - pd.DateOffset(months=pre_months))
                &
                (hist[MONTH] < event_month)
            ]

            post = hist[
                (hist[MONTH] > event_month)
                &
                (hist[MONTH] <= event_month + pd.DateOffset(months=post_months))
            ]

            if len(pre) < 2 or len(post) < 2:
                continue

            pre_avg = pre[VISITS].mean()
            post_avg = post[VISITS].mean()

            change = (
                post_avg - pre_avg
            ) / pre_avg

            events.append({

                "event_id": event_id,

                "event_type": event_type,

                "event_store_id": event_store,

                "affected_store_id": affected_store,

                "distance_miles": affected.distance,

                "submarket": affected["submarket"],

                "dma": affected["dma"],

                "opening_month": event_month,

                "pre_avg_visits": pre_avg,

                "post_avg_visits": post_avg,

                "visit_change_pct": change

            })

    return pd.DataFrame(events)
```

```python
# =========================================================
# 7. ADD YOY VISIT CHANGE
# =========================================================

def add_yoy_metrics(panel):
    panel = panel.sort_values(["our_store_id", MONTH]).copy()

    # Monthly %
    panel["mom_pct"] = (
        panel.groupby("our_store_id")["our_visits"]
        .pct_change()
    )
    
    # 3-month rolling
    
    panel["rolling3"] = (
        panel.groupby("our_store_id")["our_visits"]
        .transform(
            lambda s:
            s.rolling(3).mean()
        )
    )
    
    panel["rolling3_pct"] = (
        panel.groupby("our_store_id")["rolling3"]
        .pct_change()
    )
    
    # Quarterly
    
    panel["quarter"] = (
        panel[MONTH]
        .dt.to_period("Q")
    )

    panel["visits_lag_12"] = panel.groupby("our_store_id")["our_visits"].shift(12)

    panel["yoy_visit_change"] = (
        panel["our_visits"] - panel["visits_lag_12"]
    ) / panel["visits_lag_12"]

    panel["log_visits"] = np.log1p(panel["our_visits"])

    return panel


# =========================================================
# 8. EVENT STUDY DATASET
# =========================================================

def build_event_study_dataset(panel, store_master, geo_unit="submarket", radius=5):
    events = []

    openings = store_master[store_master["is_new_store"]].copy()

    our_stores = (
        panel[
            ["our_store_id", "our_lat", "our_lon", geo_unit]
        ]
        .drop_duplicates()
        .copy()
    )

    for _, opening in openings.iterrows():
        opening_id = opening[STORE_ID]
        opening_type = opening["store_type"]
        opening_month = opening["opening_month"]
        opening_lat = opening["lat"]
        opening_lon = opening["lon"]
        opening_geo = opening[geo_unit]

        affected = our_stores[
            (our_stores[geo_unit] == opening_geo)
            & (our_stores["our_store_id"] != opening_id)
        ].copy()

        if len(affected) == 0:
            continue

        affected["distance_miles"] = haversine_miles(
            affected["our_lat"].values,
            affected["our_lon"].values,
            opening_lat,
            opening_lon,
        )

        affected = affected[affected["distance_miles"] <= radius]

        for _, a in affected.iterrows():
            our_store_id = a["our_store_id"]

            g = panel[panel["our_store_id"] == our_store_id].copy()

            g["event_store_id"] = opening_id
            g["event_store_type"] = opening_type
            g["event_is_competitor"] = opening_type != OUR_STORE_TYPE
            g["event_is_our_store"] = opening_type == OUR_STORE_TYPE
            g["event_opening_month"] = opening_month
            g["distance_miles"] = a["distance_miles"]

            g["event_month"] = (
                (g[MONTH].dt.year - opening_month.year) * 12
                + (g[MONTH].dt.month - opening_month.month)
            )

            g = g[
                (g["event_month"] >= -EVENT_PRE_MONTHS)
                & (g["event_month"] <= EVENT_POST_MONTHS)
            ]

            events.append(g)

    if len(events) == 0:
        return pd.DataFrame()

    return pd.concat(events, ignore_index=True)


# =========================================================
# 9. EVENT STUDY SUMMARY
# =========================================================

def summarize_event_study(event_df):
    summary = (
        event_df.groupby(["event_is_competitor", "event_month"])
        .agg(
            avg_yoy_change=("yoy_visit_change", "mean"),
            median_yoy_change=("yoy_visit_change", "median"),
            avg_visits=("our_visits", "mean"),
            n=("our_store_id", "nunique"),
        )
        .reset_index()
    )

    return summary
```

```python
# =========================================================
# 10. HISTORICAL IMPACT REGRESSION
# =========================================================

def run_impact_model(panel):
    """
    Uses store fixed effects and month fixed effects.

    Target:
    yoy_visit_change

    Main variables:
    active/new competitor and our-store nearby counts.
    """

    model_df = panel.dropna(subset=["yoy_visit_change"]).copy()

    model_df["month_str"] = model_df[MONTH].dt.strftime("%Y-%m")

    formula = """
        yoy_visit_change
        ~ new_comp_3mi
        + new_comp_5mi
        + new_our_3mi
        + new_our_5mi
        + active_comp_3mi
        + active_comp_5mi
        + active_our_3mi
        + active_our_5mi
        + new_comp_weighted_score
        + new_our_weighted_score
        + C(our_store_id)
        + C(month_str)
    """

    result = smf.ols(formula=formula, data=model_df).fit(
        cov_type="cluster",
        cov_kwds={"groups": model_df["our_store_id"]}
    )

    return result
```

```python
# =========================================================
# 11. FULL PIPELINE
# =========================================================

def run_historical_impact_pipeline(
    input_csv,
    geo_unit="submarket",
    event_radius=5,
):
    df = load_store_data(input_csv)

    store_master, df_with_flags = detect_openings_and_closings(df)

    analysis_df = df_with_flags.copy()

    analysis_df = analysis_df[
        analysis_df[VISITS] > 0
    ].copy()

    panel = build_our_store_panel(analysis_df, store_master)

    panel = add_opening_features(
        panel,
        store_master,
        geo_unit=geo_unit,
    )

    panel2 = add_active_opening_features(
        panel,
        store_master,
        geo_unit=geo_unit,
    )

    active_cols = [
        c for c in panel2.columns
        if c.startswith("active_")
    ]

    panel = panel.merge(
        panel2[
            ["our_store_id", MONTH] + active_cols
        ],
        on=["our_store_id", MONTH],
        how="left",
    )

    # panel_final = add_yoy_metrics(panel_final)

    event_catalog = build_event_catalog(
        analysis_df=analysis_df,
        store_master=store_master,
        geo_unit=geo_unit,
        radius=event_radius,
        pre_months=3,
        post_months=3,
    )

    # event_summary = summarize_event_study(event_df)

    model_result = run_impact_model(panel)

    return {
        "raw_df": df,
        "analysis_df": analysis_df,
        "store_master": store_master,
        "panel": panel,
        "event_catalog": event_catalog,
        "model_result": model_result,
    }
```

```python
## Analysis - Event Catalog

event_catalog = (
    event_df.groupby(
        [
            "event_store_id",
            "our_store_id",
            "event_store_type",
            "event_is_competitor",
            "distance_miles",
        ]
    )
    .agg(
        opening_month=("event_opening_month", "first"),

        pre_visits=(
            "our_visits",
            lambda x: x[event_df.loc[x.index, "event_month"].between(-6, -1)].mean()
        ),

        post_visits=(
            "our_visits",
            lambda x: x[event_df.loc[x.index, "event_month"].between(1, 6)].mean()
        ),

        pre_yoy=(
            "yoy_visit_change",
            lambda x: x[event_df.loc[x.index, "event_month"].between(-6, -1)].mean()
        ),

        post_yoy=(
            "yoy_visit_change",
            lambda x: x[event_df.loc[x.index, "event_month"].between(1, 6)].mean()
        )
    )
    .reset_index()
)

event_catalog["visit_change_pct"] = (
    (event_catalog["post_visits"] - event_catalog["pre_visits"])
    / event_catalog["pre_visits"]
)

event_catalog["yoy_change"] = (
    event_catalog["post_yoy"] - event_catalog["pre_yoy"]
)
```

```python
## Event curve

import matplotlib.pyplot as plt

summary = (
    event_catalog
    .groupby(
        [
            "event_is_competitor",
            "event_month"
        ]
    )
    .agg(
        avg=("yoy_visit_change", "mean"),
        std=("yoy_visit_change", "std"),
        n=("our_store_id", "nunique")
    )
    .reset_index()
)

summary["se"] = summary["std"] / np.sqrt(summary["n"])
summary["lower"] = summary["avg"] - 1.96 * summary["se"]
summary["upper"] = summary["avg"] + 1.96 * summary["se"]

plt.figure(figsize=(10,5))

for flag, label in [
    (True,"Competitor Opening"),
    (False,"Our Store Opening")
]:

    g = summary[
        summary.event_is_competitor==flag
    ]

    plt.plot(
        g.event_month,
        g.avg,
        marker="o",
        label=label
    )

plt.axvline(0,color="red",linestyle="--")
plt.xlabel("Months Relative to Opening")
plt.ylabel("Average YoY Change")
plt.legend()
plt.grid(True)

plt.show()
```
